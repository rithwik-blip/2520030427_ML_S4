```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

models = {
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.01),
    "ElasticNet": ElasticNet(alpha=0.01)
}

print(models)
print(models["Ridge"])

for name, model in models.items():
    print(f"Model name: {name} -> Object: {model}")
```

**Program 7:** Loops

Write a Python program to display all elements in a list using a for loop.

Write a program to calculate the average of numbers stored in a list.

Write a program to display only the values greater than a given threshold using a loop.

Write a program to find the maximum and minimum values in a list using loops.

Write a program to count the number of elements satisfying a given condition using a loop.

```python
numbers = [10, 20, 30, 40, 50, 5, 15, 25, 35, 45]

print("1. Displaying all elements in the list:")
for num in numbers:
    print(num)

print("\n2. Calculating the average of numbers:")
total = 0
for num in numbers:
    total += num
average = total / len(numbers)
print(f"The average of the numbers is: {average}")

print("\n3. Displaying values greater than a threshold (30):")
threshold = 30
for num in numbers:
    if num > threshold:
        print(num)

print("\n4. Finding the maximum and minimum values:")
if numbers:
    maximum = numbers[0]
    minimum = numbers[0]
    for num in numbers:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    print(f"Maximum value: {maximum}")
    print(f"Minimum value: {minimum}")
else:
    print("List is empty, cannot find max/min.")

print("\n5. Counting elements greater than a threshold (20):")
count_greater_than_20 = 0
condition_threshold = 20
for num in numbers:
    if num > condition_threshold:
        count_greater_than_20 += 1
print(f"Number of elements greater than {condition_threshold}: {count_greater_than_20}")
```
